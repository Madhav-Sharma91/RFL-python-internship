```
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# Configuration
# ==========================================

INPUT_FILE = "weather_data.csv"
REPORT_FILE = "weather_report.csv"

# 1. Read CSV


df = pd.read_csv(INPUT_FILE)

df["Date"] = pd.to_datetime(df["Date"])

print("Weather data loaded successfully!")
print("\nFirst 5 records:")
print(df.head())

# 2. Average temperature for each city


average_temperature = (
    df.groupby("City")["Temperature"]
    .mean()
    .round(2)
)

print("\nAverage Temperature by City:")
print(average_temperature)
# 3. Hottest and coldest cities


hottest_city = average_temperature.idxmax()
hottest_temperature = average_temperature.max()

coldest_city = average_temperature.idxmin()
coldest_temperature = average_temperature.min()

print("\nHottest City:")
print(hottest_city, "-", hottest_temperature, "°C")

print("\nColdest City:")
print(coldest_city, "-", coldest_temperature, "°C")

# 4. Rainy and sunny days


rainy_days = (df["Weather"].str.lower() == "rainy").sum()
sunny_days = (df["Weather"].str.lower() == "sunny").sum()

print("\nRainy records:", rainy_days)
print("Sunny records:", sunny_days)

# 5. Temperature Trend


daily_temperature = (
    df.groupby("Date")["Temperature"]
    .mean()
)

plt.figure(figsize=(10, 5))

plt.plot(
    daily_temperature.index,
    daily_temperature.values,
    marker="o",
    color="red"
)

plt.title("🌡️ Temperature Trend")
plt.xlabel("Date")
plt.ylabel("Average Temperature (°C)")
plt.grid(True)

plt.tight_layout()
plt.savefig("temperature_trend.png")
plt.close()

# 6. Weather Distribution

weather_distribution = df["Weather"].value_counts()

plt.figure(figsize=(7, 5))

weather_distribution.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("🌧️ Weather Distribution")
plt.ylabel("")

plt.tight_layout()
plt.savefig("weather_distribution.png")
plt.close()

# 7. Average Temperature per City


plt.figure(figsize=(8, 5))

average_temperature.plot(
    kind="bar",
    color="orange"
)

plt.title("📊 Average Temperature per City")
plt.xlabel("City")
plt.ylabel("Average Temperature (°C)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("average_temperature_city.png")
plt.close()

# 8. Tomorrow's Temperature Prediction

daily_temperature = daily_temperature.sort_index()

if len(daily_temperature) >= 3:

    moving_average = (
        daily_temperature
        .rolling(window=3)
        .mean()
    )

    tomorrow_prediction = moving_average.iloc[-1]

    print(
        "\nTomorrow's predicted temperature:",
        round(tomorrow_prediction, 2),
        "°C"
    )

else:
    tomorrow_prediction = daily_temperature.mean()

    print(
        "\nNot enough data for a 3-day moving average."
    )


# 9. Generate Final Report


report = average_temperature.reset_index()

report.columns = [
    "City",
    "AverageTemperature"
]

report["HottestCity"] = hottest_city
report["ColdestCity"] = coldest_city

report["RainyDays"] = rainy_days
report["SunnyDays"] = sunny_days

report["TomorrowPrediction"] = round(
    tomorrow_prediction,
    2
)

report.to_csv(
    REPORT_FILE,
    index=False
)

print(
    f"\nFinal report exported to {REPORT_FILE}"
)


print("\n")
print("#*20")
print("       WEATHER SUMMARY               ")

print("Total records:", len(df))
print("Cities:", df["City"].nunique())

print(
    "Hottest city:",
    hottest_city,
    f"({hottest_temperature:.2f} °C)"
)

print(
    "Coldest city:",
    coldest_city,
    f"({coldest_temperature:.2f} °C)"
)

print("Rainy records:", rainy_days)
print("Sunny records:", sunny_days)

print(
    "Tomorrow prediction:",
    f"{tomorrow_prediction:.2f} °C"
)
```
